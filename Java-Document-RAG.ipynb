## Java Document RAG
This notebook now uses a Java PDF instead of the Think Python book. Upload `Java_Programming_Notes.pdf` or another Java PDF in the next cell. Streamlit frontend is supplied as `app.py`.

```python
from google.colab import files
uploaded = files.upload()
java_pdf = next((name for name in uploaded if name.lower().endswith(".pdf")), None)
if java_pdf is None: raise FileNotFoundError("Upload a Java PDF document")
from langchain_community.document_loaders import PyMuPDFLoader
loader = PyMuPDFLoader(java_pdf)
docs = loader.load()
print("Java document:", java_pdf, "| Pages loaded:", len(docs))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
```

```python
chunks = splitter.split_documents(docs)
```

```python
embeddings = HuggingFaceEmbeddings(model = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
from langchain_community.vectorstores import FAISS
```

```python

vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.save_local("FAISS_index")
```

Retrieval part

```python
question = "What is a class in Java?"
```

```python
results = vector_store.similarity_search_with_score(question, k=4)
```

```python
doc, score = results[0]
```

```python
print("Page:", doc.metadata.get("page", 0) + 1)
print("Distance:", score)
print("Content:")
print(doc.page_content)
```

```python
import os
from getpass import getpass
from langchain_google_genai import ChatGoogleGenerativeAI

if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass("Enter your Google Gemini API key: ")

GEMINI_MODEL = "gemini-3.5-flash"
llm = ChatGoogleGenerativeAI(model=GEMINI_MODEL, temperature=0)
print("Gemini client initialized:", GEMINI_MODEL)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """You are a helpful teaching assistant. Answer ONLY using the supplied document context.
If the answer is not present, say 'I could not find this in the provided document.'
Use simple language and cite source page numbers where possible. Do not invent citations."""
    ),
    (
        "human",
        """Question: {question}

Document context:
{context}"""
    )
])

answer_chain = prompt | llm | StrOutputParser()
```

```python
def rag_answer(question, k=4):
    retrieved = vector_store.similarity_search(question, k=k)

    context_parts = []

    for i, doc in enumerate(retrieved, 1):
        page = doc.metadata.get("page")
        page_label = str(page + 1) if isinstance(page, int) else "unknown"

        context_parts.append(
            f"[Source {i}, page {page_label}]\n{doc.page_content}"
        )

    context = "\n\n".join(context_parts)

    answer = answer_chain.invoke({
        "question": question,
        "context": context
    })

    return {
        "question": question,
        "answer": answer,
        "sources": retrieved,
        "context": context
    }
```

```python
test_questions = ["What is Java?", "Compare JDK, JRE, and JVM.", "Explain the four pillars of OOP."]
for q in test_questions:
    output = rag_answer(q)
    print("QUESTION:", q)
    print("ANSWER:", output["answer"])
    print("SOURCE PAGES:", [d.metadata.get("page", -1)+1 for d in output["sources"]])
    print("-"*70)
```
